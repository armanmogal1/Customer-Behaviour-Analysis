```python
#Filling missing values of review rating by median of each category
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median))
```

```python
df.isnull().sum()
```

```python
#making column name in lower case and replacing spaces with underscore(snake casing)
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)' : 'purchase_amount'})
```

```python
df.columns
```

```python
# creating a new column age_group
labels=['Young Adult','Adult','Middle-aged','Senior']
df['age_group'] = pd.qcut(df['age'],q=4,labels = labels)
```

```python
df.columns
```

```python
df[['age','age_group']].head(10)
```

```python
# creating column purchase_frequency_days for better analysis
frequency_mapping={
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Monthly' : 30,
    'Quarterly' : 90,
    'Bi-Weekly' : 14,
    'Annually' : 365,
    'Every 3 Months' : 90 
}

df['purchase_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
df[['discount_applied','promo_code_used']].head(10)
```

```python
#checking if both columns have same values
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
#removing promo_code_used column
df = df.drop('promo_code_used',axis=1)
```

```python
# Ye check karega har column me kaunse types hai
for col in df.columns:
    print(col, df[col].apply(type).unique())
```

```python
# Coverting method function to integer datatype
def convert_method_to_int(x):
    if callable(x):  
        try:
            return int(x(0))  
        except:
            return None
    else:
        return x

df['review_rating'] = df['review_rating'].apply(convert_method_to_int)

df['review_rating'] = df.groupby('category')['review_rating'].transform(
    lambda x: x.fillna(x.median())
)
```
